```python

# Expense Splitter
# Assignment 1 - AI Coding Assistant Practicum

def add_expense(expenses, description, amount, payer):
    """Add a new expense."""
    if amount <= 0:
        raise ValueError("Amount must be greater than zero.")

    expenses.append({
        "description": description,
        "amount": amount,
        "payer": payer
    })


def calculate_total(expenses):
    """Calculate total expenses."""
    return sum(expense["amount"] for expense in expenses)


def calculate_shares(total, participants):
    """Calculate equal share per participant."""
    if not participants:
        raise ValueError("Participants cannot be empty.")

    return total / len(participants)


def calculate_balances(expenses, participants):
    """Calculate how much each person paid and owes."""
    total = calculate_total(expenses)
    share = calculate_shares(total, participants)

    balances = {person: -share for person in participants}

    for expense in expenses:
        payer = expense["payer"]

        if payer not in balances:
            raise ValueError(f"Unknown payer: {payer}")

        balances[payer] += expense["amount"]

    return balances


def display_expenses(expenses):
    """Display all expenses."""
    for expense in expenses:
        print(
            expense["description"],
            "-", expense["amount"],
            "paid by", expense["payer"]
        )


def display_balances(balances):
    """Display final balances."""
    print("\nFinal Balances:")

    for person, balance in balances.items():
        if balance > 0:
            print(person, "should receive ₹", round(balance, 2))
        elif balance < 0:
            print(person, "owes ₹", round(abs(balance), 2))
        else:
            print(person, "is settled")


def main():
    participants = ["Asha", "Rahul", "Priya"]
    expenses = []

    add_expense(expenses, "Dinner", 900, "Asha")
    add_expense(expenses, "Taxi", 300, "Rahul")

    display_expenses(expenses)

    total = calculate_total(expenses)
    print("\nTotal Expenses: ₹", total)

    balances = calculate_balances(expenses, participants)
    display_balances(balances)


if __name__ == "__main__":
    main()
```

```python
import math


def validate_expense(expense: dict, participants: list[str]) -> bool:
    """Validate an expense dictionary against allowed participants and numeric rules."""
    if not isinstance(expense, dict):
        raise ValueError("Expense must be a dictionary.")

    amount = expense.get("amount")
    payer = expense.get("payer")

    # 1 & 4: Check type first so text or booleans fail cleanly before numeric comparisons
    if isinstance(amount, bool) or not isinstance(amount, (int, float)):
        raise ValueError("Expense amount must be a valid number.")

    if math.isnan(amount) or math.isinf(amount):
        raise ValueError("Expense amount must be a finite number.")

    # 1 & 2: Reject negative or zero amounts
    if amount <= 0:
        raise ValueError("Expense amount must be greater than zero.")

    # 3: Ensure payer is part of the group
    if payer not in participants:
        raise ValueError(f"Payer '{payer}' is not in the participant list.")

    return True
```

```python
participants = ["Asha", "Rahul", "Priya"]

# Valid expense
valid_expense = {"description": "Dinner", "amount": 900, "payer": "Asha"}
assert validate_expense(valid_expense, participants) is True

# Invalid edge cases
edge_case_expenses = [
    ("Negative amount", {"description": "Snack", "amount": -100, "payer": "Asha"}),
    ("Zero amount", {"description": "Water", "amount": 0, "payer": "Rahul"}),
    ("Unknown payer", {"description": "Taxi", "amount": 300, "payer": "Vikram"}),
    ("Text amount", {"description": "Groceries", "amount": "125.50", "payer": "Priya"}),
    ("Boolean amount", {"description": "Coffee", "amount": True, "payer": "Asha"}),
]

for label, bad_expense in edge_case_expenses:
    try:
        validate_expense(bad_expense, participants)
        print(f"[FAIL] {label}: expected ValueError")
    except ValueError as exc:
        print(f"[PASS] {label}: {exc}")
```

```python

def validate_expense(expense, participants):
    """Validate an expense before processing it."""

    amount = expense.get("amount")
    payer = expense.get("payer")

    if not isinstance(amount, (int, float)) or isinstance(amount, bool):
        raise ValueError("Expense amount must be a number.")

    if amount <= 0:
        raise ValueError("Expense amount must be greater than zero.")

    if payer not in participants:
        raise ValueError("Payer must be a valid participant.")

    return True
```
